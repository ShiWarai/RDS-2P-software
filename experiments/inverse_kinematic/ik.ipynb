```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider
from typing import Optional, Tuple, List
import math

# Константы из C++ кода
L_HIP = 10.0          # длина бедра
L_LEG = 10.255        # длина голени
L_CAROMYSLO = 3.191   # длина каромысла голени
L_CAROMYSLO_ANGLE = math.pi - (32.2 * math.pi / 180.0)  # угол каромысла в радианах
L_TYAGA = 10.0        # длина длинной тяги голени
L_SIDE_A = 2.4        # длина стороны A качалки
L_SIDE_B = 2.7        # длина стороны B качалки
L_SIDE_C_ANGLE_DEG = 90.0  # угол в градусах
L_MALA_TYAGA = 2.8    # длина малой тяги
L_RICHAG_3 = 2.4      # длина рычага двигателя 3
OFFSET_3_X = 2.0      # смещение оси двигателя 3 по X
OFFSET_3_Y = -2.0     # смещение оси двигателя 3 по Y

DEG_TO_RAD = math.pi / 180.0
RAD_TO_DEG = 180.0 / math.pi
EPSILON = 1e-10

def circle_intersection(x1: float, y1: float, r1: float, 
                       x2: float, y2: float, r2: float) -> List[Tuple[float, float]]:
    """Находит точки пересечения двух окружностей"""
    result = []
    
    d = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)
    
    # Проверяем условия пересечения
    if d > r1 + r2 + EPSILON or d < abs(r1 - r2) - EPSILON or \
       (abs(d) < EPSILON and abs(r1 - r2) < EPSILON):
        return result  # Нет пересечений
    
    a = (r1 * r1 - r2 * r2 + d * d) / (2.0 * d)
    h_sq = max(0.0, r1 * r1 - a * a)
    h = math.sqrt(h_sq)
    
    xm = x1 + a * (x2 - x1) / d
    ym = y1 + a * (y2 - y1) / d
    
    if h < EPSILON:
        # Одна точка (касание)
        result.append((xm, ym))
    else:
        # Две точки
        x3 = xm + h * (y2 - y1) / d
        y3 = ym - h * (x2 - x1) / d
        x4 = xm - h * (y2 - y1) / d
        y4 = ym + h * (x2 - x1) / d
        
        result.append((x3, y3))
        result.append((x4, y4))
    
    return result

def inverse_kinematics_for_shin_motor(motor_2_angle_rad: float, 
                                     hip_shin_angle_rad: float) -> Optional[float]:
    """
    Вычисляет угол мотора голени по углу мотора бедра и углу между бедром и голенью.
    
    Args:
        motor_2_angle_rad: угол мотора 2 (бедра) в радианах (ИНВЕРТИРОВАННЫЙ)
        hip_shin_angle_rad: угол между бедром и голенью в радианах
    
    Returns:
        угол мотора голени в радианах или None, если решение невозможно
    """
    
    # ========== ВЫЧИСЛЕНИЕ ПОЗИЦИИ КОЛЕНА ==========
    # Преобразуем угол мотора бедра в радианы и добавляем смещение 180°
    # ИНВЕРТИРУЕМ угол мотора: -motor_2_angle_rad вместо +motor_2_angle_rad
    theta_hip = (-motor_2_angle_rad + 180.0 * DEG_TO_RAD)
    
    # Колено находится на конце бедра длиной L_HIP под углом theta_hip
    knee_x = L_HIP * math.cos(theta_hip)
    knee_y = L_HIP * math.sin(theta_hip)
    
    # ========== ВЫЧИСЛЕНИЕ УГЛА ГОЛЕНИ ==========
    # Преобразуем входной угол из радиан в радианы (убираем DEG_TO_RAD)
    theta_leg = (math.pi - hip_shin_angle_rad)
    
    # ========== ВЫЧИСЛЕНИЕ ПОЗИЦИИ КАРОМЫСЛА ==========
    # Каромысло отходит от колена под углом (theta_hip + theta_leg + L_CAROMYSLO_ANGLE)
    B_ext_x = knee_x + L_CAROMYSLO * math.cos(theta_hip + theta_leg + L_CAROMYSLO_ANGLE)
    B_ext_y = knee_y + L_CAROMYSLO * math.sin(theta_hip + theta_leg + L_CAROMYSLO_ANGLE)
    
    # ========== ПОИСК ТОЧКИ D (КАЧАЛКА) ==========
    intersections = circle_intersection(0, 0, L_SIDE_A, B_ext_x, B_ext_y, L_TYAGA)
    if not intersections:
        return None  # Нет пересечений - механизм не может быть собран
    
    # Выбираем точку D с максимальной Y-координатой (верхняя точка)
    D = max(intersections, key=lambda point: point[1])
    
    # ========== ВЫЧИСЛЕНИЕ ТОЧКИ P (КАЧАЛКА) ==========
    alpha = math.atan2(D[1], D[0])
    beta = alpha - L_SIDE_C_ANGLE_DEG * DEG_TO_RAD
    
    P_x = L_SIDE_B * math.cos(beta)
    P_y = L_SIDE_B * math.sin(beta)
    
    # ========== ПОИСК ТОЧКИ R (РЫЧАГ ДВИГАТЕЛЯ 3) ==========
    intersections = circle_intersection(OFFSET_3_X, OFFSET_3_Y, L_RICHAG_3, P_x, P_y, L_MALA_TYAGA)
    if not intersections:
        return None  # Нет пересечений - механизм не может быть собран
    
    # Выбираем точку R с максимальной X-координатой (правая точка)
    R = max(intersections, key=lambda point: point[0])
    
    # ========== ВЫЧИСЛЕНИЕ УГЛА ДВИГАТЕЛЯ 3 ==========
    theta_dvigatel_3 = math.atan2(R[1] - OFFSET_3_Y, R[0] - OFFSET_3_X)
    
    # Возвращаем угол мотора голени в радианах
    return theta_dvigatel_3

def compute_positions_for_visualization(motor_2_angle_rad: float, 
                                       hip_shin_angle_rad: float) -> Optional[dict]:
    """
    Вычисляет все позиции точек для визуализации
    
    Returns:
        словарь с позициями всех точек или None, если решение невозможно
    """
    # ========== ВЫЧИСЛЕНИЕ ПОЗИЦИИ КОЛЕНА ==========
    theta_hip = (-motor_2_angle_rad + 180.0 * DEG_TO_RAD)
    knee_x = L_HIP * math.cos(theta_hip)
    knee_y = L_HIP * math.sin(theta_hip)
    
    # ========== ВЫЧИСЛЕНИЕ УГЛА ГОЛЕНИ ==========
    theta_leg = (math.pi - hip_shin_angle_rad)
    
    # ========== ВЫЧИСЛЕНИЕ ПОЗИЦИИ КАРОМЫСЛА ==========
    B_ext_x = knee_x + L_CAROMYSLO * math.cos(theta_hip + theta_leg + L_CAROMYSLO_ANGLE)
    B_ext_y = knee_y + L_CAROMYSLO * math.sin(theta_hip + theta_leg + L_CAROMYSLO_ANGLE)
    
    # ========== ПОИСК ТОЧКИ D (КАЧАЛКА) ==========
    intersections = circle_intersection(0, 0, L_SIDE_A, B_ext_x, B_ext_y, L_TYAGA)
    if not intersections:
        return None
    
    D = max(intersections, key=lambda point: point[1])
    
    # ========== ВЫЧИСЛЕНИЕ ТОЧКИ P (КАЧАЛКА) ==========
    alpha = math.atan2(D[1], D[0])
    beta = alpha - L_SIDE_C_ANGLE_DEG * DEG_TO_RAD
    P_x = L_SIDE_B * math.cos(beta)
    P_y = L_SIDE_B * math.sin(beta)
    
    # ========== ПОИСК ТОЧКИ R (РЫЧАГ ДВИГАТЕЛЯ 3) ==========
    intersections = circle_intersection(OFFSET_3_X, OFFSET_3_Y, L_RICHAG_3, P_x, P_y, L_MALA_TYAGA)
    if not intersections:
        return None
    
    R = max(intersections, key=lambda point: point[0])
    
    # ========== ВЫЧИСЛЕНИЕ ПОЗИЦИИ СТУПНИ ==========
    # Ступня находится на конце голени от колена
    foot_x = knee_x + L_LEG * math.cos(theta_hip + theta_leg)
    foot_y = knee_y + L_LEG * math.sin(theta_hip + theta_leg)
    
    return {
        'origin': (0, 0),
        'knee': (knee_x, knee_y),
        'B_ext': (B_ext_x, B_ext_y),
        'D': D,
        'P': (P_x, P_y),
        'R': R,
        'foot': (foot_x, foot_y),
        'motor3_offset': (OFFSET_3_X, OFFSET_3_Y)
    }

def plot_lever_system_inverse(motor_2_angle_rad: float, hip_shin_angle_rad: float):
    """
    Визуализация кинематической цепи робота на основе углов двигателей.
    
    Args:
        motor_2_angle_rad: Угол двигателя 2 (бедра) в радианах
        hip_shin_angle_rad: Угол между бедром и голенью в радианах
    """
    motor_2_angle_rad -= math.pi/4
    hip_shin_angle_rad += math.pi/2
    # Вычисляем угол мотора голени
    shin_motor_angle_rad = inverse_kinematics_for_shin_motor(motor_2_angle_rad, hip_shin_angle_rad)
    
    if shin_motor_angle_rad is None:
        print("❌ Невозможно найти решение для данных углов!")
        return
    
    # Получаем позиции всех точек
    positions = compute_positions_for_visualization(motor_2_angle_rad, hip_shin_angle_rad)
    
    if positions is None:
        print("❌ Невозможно вычислить позиции точек!")
        return
    
    print(f"✅ Угол мотора бедра: {motor_2_angle_rad:.3f} рад ({motor_2_angle_rad * RAD_TO_DEG:.2f}°)")
    print(f"✅ Угол между бедром и голенью: {hip_shin_angle_rad:.3f} рад ({hip_shin_angle_rad * RAD_TO_DEG:.2f}°)")
    print(f"✅ Угол мотора голени: {shin_motor_angle_rad:.3f} рад ({shin_motor_angle_rad * RAD_TO_DEG:.2f}°)")
    
    # Извлекаем позиции
    origin = positions['origin']
    knee = positions['knee']
    B_ext = positions['B_ext']
    D = positions['D']
    P = positions['P']
    R = positions['R']
    foot = positions['foot']
    motor3_offset = positions['motor3_offset']
    
    # Создаем график
    plt.figure(figsize=(12, 10))
    
    # Основные звенья ноги
    plt.plot([origin[0], knee[0]], [origin[1], knee[1]], '-o', color='blue', linewidth=3, label='Бедро')
    plt.plot([knee[0], foot[0]], [knee[1], foot[1]], '-o', color='green', linewidth=3, label='Голень')
    
    # Каромысло и тяги
    plt.plot([knee[0], B_ext[0]], [knee[1], B_ext[1]], '-o', color='orange', linewidth=2, label='Каромысло голени')
    plt.plot([B_ext[0], D[0]], [B_ext[1], D[1]], '-o', color='purple', linewidth=2, label='Длинная тяга')
    
    # Качалка
    plt.plot([origin[0], D[0]], [origin[1], D[1]], '-o', color='red', linewidth=2, label='Сторона A качалки')
    plt.plot([origin[0], P[0]], [origin[1], P[1]], '-o', color='red', linewidth=2, label='Сторона B качалки')
    plt.plot([D[0], P[0]], [D[1], P[1]], '-o', color='red', linewidth=2, label='Сторона C качалки')
    
    # Малая тяга и рычаг двигателя 3
    plt.plot([P[0], R[0]], [P[1], R[1]], '-o', color='cyan', linewidth=2, label='Малая тяга')
    plt.plot([motor3_offset[0], R[0]], [motor3_offset[1], R[1]], '-o', color='magenta', linewidth=2, label='Рычаг двигателя 3')
    
    # Ключевые точки
    plt.plot(origin[0], origin[1], 'ko', markersize=8, label='Начало координат')
    plt.plot(knee[0], knee[1], 'bo', markersize=8, label='Колено')
    plt.plot(foot[0], foot[1], 'go', markersize=8, label='Ступня')
    plt.plot(motor3_offset[0], motor3_offset[1], 'ro', markersize=8, label='Ось двигателя 3')
    
    # Настройка графика
    plt.xlim(-15, 15)
    plt.ylim(-15, 15)
    plt.axhline(0, color='black', linewidth=0.5)
    plt.axvline(0, color='black', linewidth=0.5)
    plt.grid(color='gray', linestyle='--', linewidth=0.5)
    plt.gca().set_aspect('equal', adjustable='box')
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.title(f'Кинематика робота-ноги\nУгол бедра: {motor_2_angle_rad:.3f} рад, Угол бедро-голень: {hip_shin_angle_rad:.3f} рад')
    plt.xlabel('X (см)')
    plt.ylabel('Y (см)')
    plt.tight_layout()
    plt.show()

# Интерактивная визуализация
print("🎮 Интерактивная визуализация кинематики робота-ноги")
print("🎛️ Используйте слайдеры для изменения углов в радианах:")
print("   • Угол двигателя 2 (бедра): -π до π радиан")
print("   • Угол между бедром и голенью: 0 до π радиан")
print("   • Угол мотора 2 инвертируется внутри функции")
print("   • Все вычисления выполняются в радианах")

interact(
    plot_lever_system_inverse,
    motor_2_angle_rad=FloatSlider(min=-math.pi, max=math.pi, step=0.01, value=0, description='Угол мотора бедра (рад)'),
    hip_shin_angle_rad=FloatSlider(min=0, max=math.pi, step=0.01, value=math.pi/4, description='Угол бедро-голень (рад)')
)
```
